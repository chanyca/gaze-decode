```python
import pandas as pd
from matplotlib import pyplot as plt
import numpy as np
import seaborn as sns
import sys
import os
import importlib

sys.path.append(os.path.abspath('EngbertMicrosaccadeToolbox'))
from EngbertMicrosaccadeToolbox import microsac_detection
importlib.reload(microsac_detection)
# import microsac_detection
```

```python
d = pd.read_csv('eyelink_csv/SV009_sample_trialno_phase.csv')
d.head()
# Get rows for trial_no=1, phase='s', LX, LY not empty
# d = d[(d['trial_no']==400) & (d['phase']=='s') & (d['LX'].notnull()) & (d['LY'].notnull())]
d = d[(d['phase']=='s') & (d['LX'].notnull()) & (d['LY'].notnull())]
# print(d.head())
# print(d.trial_no.unique())

# plt.plot(d.LX, label='LX')
# plt.plot(d.LY, label='LY')
# plt.legend()

# Set parameters
SAMPLING = 250 # sampling rate, hz
MINDUR = 1 # minimal saccade duration, originally 3
VFAC = 5 # relative velocity threshold, originally 5

microsac_dict = {}
for trial_no in d.trial_no.unique():
    trial_d = d[d['trial_no']==trial_no]
    # print(trial_d.head())
    left_eye = np.vstack([np.asarray(trial_d.LX), np.asarray(trial_d.LY)]).T
    if left_eye.size > 0:
        ms_l, rad_l = microsac_detection.microsacc(left_eye, vfac=VFAC, mindur=MINDUR, sampling=SAMPLING)
        microsac_dict[trial_no] = ms_l
    else:
        print(f"Skipping trial_no {trial_no} due to empty left_eye array")

microsac_dict
```
